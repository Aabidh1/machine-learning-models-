### Data cleaning and preprocessing

```python
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
print("Numerical Columns:", list(numerical_cols))
```

```python
df = df.drop(["EmployeeCount","EmployeeNumber","Over18","StandardHours","HourlyRate", "DailyRate"], axis=1)
```

```python
df["Attrition"] = df["Attrition"].map({"Yes":1, "No":0})
```

```python
df["OverTime"] = df["OverTime"].map({"Yes":1, "No":0})
```

```python
numeric_df = df.select_dtypes(include=['number'])
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder

df_encoded = df.copy()
le = LabelEncoder()
for col in df_encoded.select_dtypes(include="object"):
    df_encoded[col] = le.fit_transform(df_encoded[col])

X = df_encoded.drop("Attrition", axis=1)
y = df_encoded["Attrition"]

model = RandomForestClassifier(random_state=42)
model.fit(X, y)

import pandas as pd
fi = pd.DataFrame({"Feature":X.columns, "Importance":model.feature_importances_})
print(fi.sort_values("Importance", ascending=False).head(10))
```

```python
df_no_outliers = df_selected.copy()

numeric_features = df_no_outliers.select_dtypes(include=['int64', 'float64']).columns.tolist()

numeric_features = [col for col in numeric_features if col not in ['Attrition', 'OverTime']]

def remove_outliers(col):
    Q1 = col.quantile(0.25)
    Q3 = col.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return col[(col >= lower) & (col <= upper)]

df_no_outliers = df_no_outliers[numeric_features].apply(remove_outliers).dropna()

df_no_outliers['Attrition'] = df_selected.loc[df_no_outliers.index, 'Attrition']
df_no_outliers['OverTime'] = df_selected.loc[df_no_outliers.index, 'OverTime']

print(df_no_outliers.head())
```

```python
from sklearn.preprocessing import StandardScaler

df_scaled = df_no_outliers.copy()

numeric_features = df_scaled.select_dtypes(include=['int64', 'float64']).columns.tolist()
numeric_features = [col for col in numeric_features if col not in ['Attrition', 'OverTime']]

scaler = StandardScaler()

df_scaled[numeric_features] = scaler.fit_transform(df_scaled[numeric_features])

print(df_scaled.head())
```

### Model Building Testing

```python
X = df_scaled.drop("Attrition", axis=1)
y = df_scaled["Attrition"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=42)
```

```python
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)

y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)
```

### Model Evaluation

```python
print("=== Training Set Evaluation ===")
print("Accuracy:", accuracy_score(y_train, y_train_pred))
```

```python
print("\n=== Testing Set Evaluation ===")
print("Accuracy:", accuracy_score(y_test, y_test_pred))
```
